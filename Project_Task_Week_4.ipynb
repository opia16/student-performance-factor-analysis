**Visualisasi Data**

```python
df_corr = df.copy()
df_corr['Extracurricular Activities'] = df_corr['Extracurricular Activities'].map({'Yes': 1, 'No': 0})

plt.figure(figsize=(10, 8))
sns.heatmap(df_corr.corr(), annot=True, cmap='coolwarm', fmt=".2f")
plt.title('Correlation Matrix')
plt.show()
```

**Insight:**
Berdasarkan Statistical Profiling, Performance Index memiliki rata-rata 55 dengan standar deviasi ±19, menunjukkan variasi performa yang sangat tinggi antar siswa. Ketika dikombinasikan dengan temuan Visual Exploration berupa heatmap korelasi, terungkap bahwa variabel Previous Scores memiliki korelasi sangat kuat terhadap Performance Index dengan nilai 0.92 jauh melampaui variabel lainnya.
Dalam konteks bisnis EdTech, angka 0.92 adalah sinyal bahaya: sistem pembelajaran yang ada cenderung mempertahankan kesenjangan, bukan mempersempitnya. Siswa yang sudah unggul secara historis akan terus unggul, sementara siswa yang pernah tertinggal nyaris tidak memiliki ruang untuk mengejar ketertinggalan hanya dengan pola belajar yang sama.


**Hipotesis:** Nilai Sebelumnya memiliki pengaruh signifikan terhadap Performance Index, di mana siswa dengan nilai sebelumnya yang tinggi cenderung memiliki performa yang tinggi pula. Diperlukan uji beda rata-rata (independent t-test) untuk membuktikan apakah peningkatan Hours Studied memberikan dampak yang berbeda secara statistik antara kelompok siswa dengan Previous Scores rendah (< 50) vs tinggi (≥ 50).


```python
plt.figure(figsize=(8,5))

sns.regplot(data=df,
            x='Previous Scores',
            y='Performance Index',
            scatter=False,
            color='red')

plt.title('Hubungan Nilai Sebelumnya terhadap Performance Index')
plt.xlabel('Nilai Sebelumnya')
plt.ylabel('Performance Index')

plt.show()
```

**Insight:**
Berdasarkan hasil visualisasi, terdapat hubungan linear positif yang sangat kuat antara Nilai Sebelumnya dan Performance Index. Hal ini terlihat dari pola sebaran data yang mengikuti tren meningkat, di mana siswa dengan nilai sebelumnya yang tinggi cenderung memiliki Performance Index yang lebih tinggi.
Temuan ini menunjukkan bahwa performa akademik siswa bersifat relatif konsisten, sehingga kemampuan awal menjadi faktor utama dalam menentukan hasil belajar saat ini. Dengan kata lain, siswa yang telah memiliki dasar akademik yang baik cenderung mampu mempertahankan bahkan meningkatkan performanya.


**Hipotesis:**
Terdapat hubungan positif yang signifikan antara Nilai Sebelumnya dan Performance Index, di mana peningkatan nilai sebelumnya akan diikuti dengan peningkatan performa siswa. Siswa dengan Nilai Sebelumnya di bawah median (< 50) tidak akan merespons program intervensi standar secara signifikan. Karena intervensi standar dirancang untuk siswa yang sudah siap belajar lebih keras. Siswa dengan Nilai Sebelumnya rendah membutuhkan sesuatu yang berbeda: perbaikan fondasi konsep, pendekatan yang membangun kepercayaan diri, dan target yang lebih realistis secara bertahap.



```python
plt.figure(figsize=(8,5))

sns.regplot(data=df, x='Hours Studied',
            y='Performance Index',
            scatter_kws={'color':'red'},
            line_kws={'color':'pink'})

plt.title("Hubungan Jam Belajar terhadap Performance Index", color='white')
plt.xlabel("Jam Belajar", color='white')
plt.ylabel("Performance Index", color='white')

plt.show()
```

 **Insight:** ecara statistik, rata-rata jam belajar siswa adalah 5 jam dengan median 5 jam, distribusi yang sangat simetris dan menunjukkan tidak ada skewness ekstrem. Artinya, mayoritas siswa sudah berada di kisaran durasi belajar yang sama. Namun Visual Exploration melalui scatter plot Hubungan Jam Belajar terhadap Performa Index mengungkap sesuatu yang kontradiktif: meskipun garis regresi menunjukkan tren positif, penyebaran titik data sangat lebar di sepanjang garis tersebut.
Ini berarti ada siswa yang belajar 9–10 jam per hari namun tetap menghasilkan Performance Index yang rendah, sekaligus ada siswa yang hanya belajar 2–3 jam namun performanya tinggi. Fenomena ini mengindikasikan bahwa kuantitas jam belajar bukan satu-satunya penentu hasil, ada variabel moderator yang belum tertangkap.

**Hipotesis:** Terdapat hubungan positif yang signifikan antara Hours Studied dan Performance Index, di mana siswa yang belajar lebih lama cenderung memiliki Performance Index yang lebih tinggi.

```python
plt.figure(figsize=(8,5))
sns.boxplot(x='Extracurricular Activities', y='Performance Index', data=df)

plt.title("Perbandingan Performance Index Berdasarkan Ekstrakurikuler")
plt.xlabel("Ekstrakurikuler")
plt.ylabel("Performance Index")

plt.show()
```

**Insight:** Boxplot perbandingan Performance Index berdasarkan keikutsertaan ekstrakurikuler menunjukkan bahwa rata-rata performa siswa yang aktif ekstrakurikuler sedikit lebih tinggi. Namun secara statistik deskriptif, rentang sebaran (IQR) kedua kelompok sangat tumpang tindih (overlap), menandakan perbedaan ini kemungkinan tidak signifikan secara statistik.
Yang lebih menarik dari perspektif bisnis adalah interpretasi di balik temuan ini. Keikutsertaan dalam ekstrakurikuler bisa menjadi proxy untuk karakter siswa yang lebih terstruktur, mereka yang mampu mengelola waktu antara akademik dan kegiatan lain. Artinya, bukan ekstrakurikulernya yang meningkatkan performa, melainkan profil siswa yang memilih ikut ekstrakurikuler cenderung juga memiliki manajemen waktu dan motivasi belajar yang lebih baik.


**Hipotesis:** Tidak ada perbedaan signifikan dalam Performance Index antara siswa yang mengikuti kegiatan ekstrakurikuler dan siswa yang tidak mengikuti kegiatan ekstrakurikuler. Meskipun ada sedikit perbedaan rata-rata, faktor-faktor lain kemungkinan memiliki pengaruh yang lebih besar terhadap Performance Index.
